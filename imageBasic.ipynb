```python
import cv2
from pathlib import Path
```

```python
image_folder = Path("dataset/bottle/train/good")
```

```python
images = list(image_folder.glob("*.png"))
```

```python
print("Number of images:", len(images))
```

```python
image_path = images[0]
```

```python
image = cv2.imread(image_path)
```

```python
if image is None :
    print('Imgae not Loaded')
else:
    print('Image Loaded')
```

```python
    print("Shape:", image.shape)
    print("Dimensions:", image.ndim)
    print("Data type:", image.dtype)
```

```python
from matplotlib import pyplot as plt
```

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image_rgb)
plt.title("Industrial Bottle")
plt.axis("off")
plt.show()
```

```python
plt.imshow(image)
plt.title("Industrial Bottle")
plt.axis("off")
plt.show()
```

```python
print("Image shape:", image.shape)
print("Number of dimensions:", image.ndim)
print("Data type:", image.dtype)
print("Minimum pixel value:", image.min())
print("Maximum pixel value:", image.max())
print("Average pixel value:", image.mean())

print("First pixel:", image[0, 0])
```

```python
blue = image[:, :, 0]
green = image[:, :, 1]
red = image[:, :, 2]

print("Blue channel shape:", blue.shape)
print("Green channel shape:", green.shape)
print("Red channel shape:", red.shape)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(blue, cmap="gray")
plt.title("Blue Channel")

plt.subplot(1, 3, 2)
plt.imshow(green, cmap="gray")
plt.title("Green Channel")

plt.subplot(1, 3, 3)
plt.imshow(red, cmap="gray")
plt.title("Red Channel")

plt.show()
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(image[:, :, 0], cmap="Blues")
plt.title("Blue Channel")

plt.subplot(1, 3, 2)
plt.imshow(image[:, :, 1], cmap="Greens")
plt.title("Green Channel")

plt.subplot(1, 3, 3)
plt.imshow(image[:, :, 2], cmap="Reds")
plt.title("Red Channel")

plt.show()
```
